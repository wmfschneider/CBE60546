### Q1.3. What size PFR would you need, and what would be the residence time in the reactor?

For a PFR with constant $\nu$:

$
\nu \frac{dC_A}{dV} = -r_A = -k C_A^2
$

$
\frac{dC_A}{C_A^2} = -\frac{k}{\nu}\,dV
$

Integrate from inlet $C_{A0}$ to outlet $C_A$:

$
\int_{C_{A0}}^{C_A} \frac{dC_A}{C_A^2}
= -\frac{k}{\nu}\int_0^{V} dV
$

$
\left[-\frac{1}{C_A}\right]_{C_{A0}}^{C_A}
= -\frac{kV}{\nu}
\quad\Rightarrow\quad
\frac{1}{C_A} - \frac{1}{C_{A0}} = \frac{kV}{\nu}
$

$
V = \frac{\nu}{k}\left(\frac{1}{C_A} - \frac{1}{C_{A0}}\right)
$

Insert numbers:

$
V = \frac{0.922}{0.5}
    \left(\frac{1}{0.020} - \frac{1}{0.444}\right)\ \text{L}
  \approx 8.8\times 10^{1}\ \text{L}
  = 8.8\times 10^{-2}\ \text{m}^3
$

Residence time (space time):

$
\tau = \frac{V}{\nu}
     = \frac{88.1\ \text{L}}{0.922\ \text{L/s}}
     \approx 95.5\ \text{s}
     \approx 1.6\ \text{min}
$

### Q1.4. A "Levenspiel" plot is a plot of $1/r$ vs conversion of a reactant. Create a Levenspiel plot to rationalize your answers above.

$\frac{1}{r} = \frac{1}{-r_A}$ where $r_A = k C_A C_B = k C_A^2 = -k C_{A0}^2 (1-X)^2$

so $ \frac{1}{r} = \frac{1}{k C_{A0}^2 (1-X)^2} = fn(X)$

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
k   = 0.5      # L/(mol·s)
CA0 = 0.4444   # mol/L  (mixed inlet concentration of A)

# Rate as a function of conversion X
def rA(X):
    return k * CA0**2 * (1 - X)**2

# Conversion grid
X = np.linspace(0.0, 0.97, 300)
rA_vals   = rA(X)
inv_rA    = 1.0 / rA_vals   # (L·s/mol)

# Design conversion
X_target = 0.955
rA_target    = rA(X_target)
inv_rA_target = 1.0 / rA_target

plt.figure()

# Curve: 1/(-rA)
plt.plot(X, inv_rA, label=r"$1/(-r_A)$", color='blue')

# Vertical line at target conversion
plt.axvline(X_target, linestyle=":", color="gray")

# Horizontal line at CSTR height
plt.axhline(inv_rA_target, linestyle="--", color="gray")

# Shade the CSTR rectangle
plt.fill_between(
    [0, X_target],
    [inv_rA_target, inv_rA_target],
    [0, 0],
    alpha=0.25,
    color='orange',
    label="CSTR volume (rectangle)"
)

# Shade the PFR area under the curve
mask = X <= X_target
plt.fill_between(
    X[mask],
    inv_rA[mask],
    np.zeros_like(inv_rA[mask]),
    alpha=0.25,
    color='green',
    label="PFR volume (area under curve)"
)

# Mark the point at X_target
plt.scatter([X_target], [inv_rA_target], color="black", zorder=3)

plt.xlabel("Conversion of A, X")
plt.ylabel(r"$1/(-r_A)$  (L·s/mol)")
plt.legend()
plt.tight_layout()
plt.show()
```

Because the CSTR volume corresponds to the rectangular area at the exit conversion while the PFR volume corresponds to the area under the $1/(-r_A)$ curve, the rectangle is visibly larger than the shaded PFR area. This geometric comparison directly shows that the CSTR requires a larger reactor volume than the PFR to achieve the same conversion.

### Q1.6. The largest CSTR you can find is 1000 l in volume. At what temperature would you have to operate the reactor to achieve your production goal? The apparent activation energy of the reaction is 50 kJ/mol and you can assume the densities are independent of temperature.

From Q1.2, at **25°C (298 K)** the CSTR volume needed to meet the
production and conversion target was

$
V_{25} \approx 1.96 \,\text{m}^3 = 1960\ \text{L}
$

But the largest available reactor is

$
V_{\text{avail}} = 1000\ \text{L} = 1.0\ \text{m}^3
$

For a CSTR with fixed inlet/outlet concentrations and volumetric flow rate
(densities independent of T), the design equation

$
V = \frac{\dot V (C_{A0}-C_A)}{k\,C_A C_B}
$

shows that

$
V \propto \frac{1}{k}
\quad\Rightarrow\quad
\frac{k_{\text{new}}}{k_{25}} = \frac{V_{25}}{V_{\text{avail}}}
$

Using $V_{25} \approx 1.87\ \text{m}^3$ and $V_{\text{avail}} = 1.0\ \text{m}^3$:

$
\frac{k_{\text{new}}}{k_{25}} \approx 1.96
$

---

### Arrhenius relation

Given:

- $k_{25} = 0.5\ \text{M}^{-1}\text{s}^{-1}$ at $T_{25} = 298\ \text{K}$,
- Activation energy $E_a = 50\ \text{kJ/mol} = 5.0\times 10^4\ \text{J/mol}$,
- Gas constant $R = 8.314\ \text{J/mol·K}$.

Arrhenius equation:

$
\frac{k_{\text{new}}}{k_{25}}
= \exp\left[-\frac{E_a}{R}\left(\frac{1}{T_{\text{new}}} - \frac{1}{T_{25}}\right)\right]
$

Set $k_{\text{new}}/k_{25} = 1.96$:

$
\ln(1.96) = -\frac{E_a}{R}\left(\frac{1}{T_{\text{new}}} - \frac{1}{T_{25}}\right)
$

$
\frac{1}{T_{\text{new}}}
= \frac{1}{T_{25}} - \frac{R}{E_a}\ln(1.96)
$

Plugging in numbers:

$
\frac{1}{T_{\text{new}}}
\approx \frac{1}{298\ \text{K}} - \frac{8.314}{5.0\times10^4}\ln(1.96)
$

$
T_{\text{new}} \approx 3.08\times10^2\ \text{K}
\approx \boxed{308\ \text{K}}
$

### Q1.7 (BONUS) A bright colleague suggests that the reaction could be run more effectively in a gas-phase, isobaric (constant pressure), isothermal PFR with no diluent. The gas-phase reaction follows the same rate law and has the same rate constant at $200^\circ$ that the aqueous reaction has at $25^\circ$. What size PFR would you need and at what space velocity would you need to operate it? You can treat all the gases as ideal. (*Hint* Because the reaction takes two molecules to one, the volumetric flow rate must increase down the reactor to maintain constant pressure. You will need to incorporate this change in the calculation of concentration as reaction progresses.)

### 1. Feed molar flow and total concentration

Product rate (from Q1.1):

$
\dot n_P = \frac{100\,000\ \text{g/h}}{71.08\ \text{g/mol}}
         = 1.41\times10^3\ \text{mol/h}
         = 3.91\times10^{-1}\ \text{mol/s}
$

Using $X = 0.955$,

$
F_{A0} = F_{B0} = \frac{\dot n_P}{X}
        \approx \frac{0.391}{0.955}
        \approx 0.409\ \text{mol/s}
$

Stoichiometric feed → $F_{T0} = F_{A0}+F_{B0} = 2 F_{A0}$.

Assume ideal gas at **1 atm** and **200°C = 473 K**:

$
C_T = \frac{P}{RT}
     = \frac{1.013\times10^5}{8.314 \times 473}
     \approx 25.8\ \text{mol/m}^3
$

Rate constant at 200°C: same numerical value as liquid case,  
$k = 0.5\ \text{L mol}^{-1}\text{s}^{-1} = 5.0\times10^{-4}\ \text{m}^3\text{mol}^{-1}\text{s}^{-1}$.

---

### 2. Concentrations and rate as a function of conversion

For equimolar, no-diluent gas feed and reaction  
$\text{A}+\text{B}\to\text{P}$:

- $F_A = F_{A0}(1-X)$
- $F_B = F_{B0}(1-X)$
- $F_T = F_A+F_B+F_P = F_{A0}(2-X)$

Mole fraction of A:

$
y_A = \frac{F_A}{F_T} = \frac{1-X}{2-X}
$

At constant T, P: $C_A = y_A C_T$, and because $C_A = C_B$:

$
r = k C_A C_B = k C_A^2
   = k C_T^2\left(\frac{1-X}{2-X}\right)^2
$

---

### 3. PFR design equation

Mole balance on A in a PFR:

$
\frac{dF_A}{dV} = r_A = -k C_T^2 \left(\frac{1-X}{2-X}\right)^2
$

But $F_A = F_{A0}(1-X)$, so

$
\frac{dF_A}{dV} = -F_{A0}\frac{dX}{dV}
$

$
F_{A0} \frac{dX}{dV}
  = k C_T^2 \frac{(1-X)^2}{(2-X)^2}
$

$
dV = \frac{F_{A0}}{k C_T^2}\,\frac{(2-X)^2}{(1-X)^2}\,dX
$

Integrate from $X=0$ to $X=X_f = 0.955$:

$
V = \frac{F_{A0}}{k C_T^2}
    \int_0^{X_f} \frac{(2-X)^2}{(1-X)^2}\,dX
$

The integral evaluates to

$
\int_0^{X_f} \frac{(2-X)^2}{(1-X)^2}\,dX
= X_f - 1 - \frac{1}{X_f-1} - 2\ln(1-X_f)
$

Numerically (with $X_f = 0.955$):

$
\int_0^{X_f} \frac{(2-X)^2}{(1-X)^2}\,dX \approx 28.38
$

Thus

$
V = \frac{0.409\ \text{mol/s}}{(5.0\times10^{-4}\ \text{m}^3\text{mol}^{-1}\text{s}^{-1})
         (25.8\ \text{mol/m}^3)^2}
    \times 28.38
  \approx 3.5\times10^{1}\ \text{m}^3
$

$
\boxed{V \approx 3.5\times10^{1}\ \text{m}^3 \; \approx 3.5\times10^{4}\ \text{L}}
$

### 4. Space velocity

Inlet volumetric flow at 200°C and 1 atm:

$

$

The **space time** (based on inlet conditions) is

$

Space velocity (SV) is the inverse:

$
\boxed{

$

So the gas-phase PFR would actually be **much larger** than the aqueous PFR you designed earlier.

### Q1.8. The PFR idea was a bust, your colleague was fired, and you get the budget to buy the CSTR you want. You propose to fill the reactor with water before starting to flow in reactants. How long will you have to wait for effluent concentrations to reach 90% of steady-state?

### 1. Mole balance for A

Moles of A in the reactor: $N_A = C_A V$.

Dynamic balance:

$
\frac{dN_A}{dt} = F_{A0} - F_A - V k C_A C_B
$

Using $F_{A0} = \nu C_{A0}$, $F_A = \nu C_A$, and constant volume $V$:

$
V \frac{dC_A}{dt} = \nu C_{A0} - \nu C_A - V k C_A C_B
$

Divide by $V$ and define the space time $\tau = V/\nu$ so that $\nu/V = 1/\tau$:

$
\frac{dC_A}{dt} = \frac{C_{A0} - C_A}{\tau} - k C_A C_B
$

### 2. Using symmetry: $C_A(t) = C_B(t)$

The mole balance for B is:

$
V \frac{dC_B}{dt} = \nu C_{B0} - \nu C_B - V k C_A C_B
$

Given:

- Stoichiometric equimolar feed: $C_{A0} = C_{B0}$
- Initially reactor is filled with water: $C_A(0) = C_B(0) = 0$
- The two ODEs for $C_A$ and $C_B$ are symmetric

By symmetry and uniqueness of ODE solutions:

$
C_A(t) = C_B(t) \quad \forall t
$

Therefore, set $C_B = C_A$ in the A-balance:

$
\frac{dC_A}{dt} = \frac{C_{A0} - C_A}{\tau} - k C_A^2
$

This is now a single nonlinear ODE for $C_A(t)$.

Rewrite:

$
\frac{dC_A}{dt} = -k C_A^2 - \frac{1}{\tau}C_A + \frac{C_{A0}}{\tau}
$

### 3. Steady states and factorization

Steady state satisfies:

$
0 = -k C_A^2 - \frac{1}{\tau}C_A + \frac{C_{A0}}{\tau}
$

Multiply by $-1$:

$
k C_A^2 + \frac{1}{\tau}C_A - \frac{C_{A0}}{\tau} = 0
$

This quadratic has two roots $C_1$ and $C_2$:

$
C_{1,2} = \frac{-\frac{1}{\tau} \pm
\sqrt{\left(\frac{1}{\tau}\right)^2 + 4 k \frac{C_{A0}}{\tau}}}{2k}
$

- $C_1$ is the **physical steady-state concentration** (positive, e.g. $0.02\ \text{M}$).
- $C_2$ is a negative, unphysical root (but useful mathematically).

We can factor the RHS as:

$
\frac{dC_A}{dt}
= -k (C_A - C_1)(C_A - C_2)
$

### 4. Separation of variables and integration

Separate variables:

$
\frac{dC_A}{(C_A - C_1)(C_A - C_2)} = -k\,dt
$

Use partial fractions:

$
\frac{1}{(C_A - C_1)(C_A - C_2)}
= \frac{1}{C_1 - C_2}\left(\frac{1}{C_A - C_1} - \frac{1}{C_A - C_2}\right)
$

Integrate:

$
\int \frac{dC_A}{(C_A - C_1)(C_A - C_2)}
= \frac{1}{C_1 - C_2} \ln\left(\frac{C_A - C_1}{C_A - C_2}\right)
$

So the ODE gives:

$
\frac{1}{C_1 - C_2} \ln\left(\frac{C_A - C_1}{C_A - C_2}\right)
= -k t + \text{const}
$

Apply the initial condition \(C_A(0) = 0\):

$
\frac{1}{C_1 - C_2} \ln\left(\frac{0 - C_1}{0 - C_2}\right)
= \frac{1}{C_1 - C_2} \ln\left(\frac{C_1}{C_2}\right)
= \text{const}
$

Thus:

$
\frac{1}{C_1 - C_2} \ln\left(\frac{C_A - C_1}{C_A - C_2}\right)
= -k t + \frac{1}{C_1 - C_2} \ln\left(\frac{C_1}{C_2}\right)
$

Multiply by \(C_1 - C_2\) and exponentiate:

$
\ln\left(\frac{C_A - C_1}{C_A - C_2}\right)
= -k (C_1 - C_2) t + \ln\left(\frac{C_1}{C_2}\right)
$

$
\frac{C_A - C_1}{C_A - C_2}
= \frac{C_1}{C_2}\,\exp\!\big[-k (C_1 - C_2)t\big]
$

### 5. Solve explicitly for $C_A(t)$

Let
$
\gamma = \frac{C_1}{C_2}, \quad
\Delta = C_1 - C_2, \quad
E(t) = \exp[-k \Delta t]
$

Then:

$
\frac{C_A - C_1}{C_A - C_2} = \gamma E(t)
$

$
C_A - C_1 = \gamma E(t)(C_A - C_2)
$

Collect \(C_A\) terms:

$
C_A - \gamma E(t) C_A = C_1 - \gamma E(t) C_2
$

$
C_A \big[1 - \gamma E(t)\big]
= C_1 - \gamma E(t) C_2
$

But \(\gamma C_2 = C_1\), so:

$
C_1 - \gamma E(t) C_2 = C_1 - C_1 E(t) = C_1 [1 - E(t)]
$

Thus the explicit solution is:

$
\boxed{
C_A(t)
= \frac{C_1\,[1 - e^{-k(C_1 - C_2)t}]}
       {1 - \dfrac{C_1}{C_2} e^{-k(C_1 - C_2)t}}
}
$

where:

- $C_1$ is the **physical steady-state concentration** of A (e.g. $0.02\ \text{M}$),
- $C_2$ is the negative root of the quadratic,
- and $C_A(0) = 0$.

As $t \to \infty$, the exponential term vanishes and$C_A(t) \to C_1$, as required.

### 6. Time to reach 90% of steady state

To find the time $t_{90}$ such that

$
C_A(t_{90}) = 0.9\,C_1
$

you substitute $C_A = 0.9 C_1$ into the expression above and solve for $t_{90}$:

$
0.9 C_1
= \frac{C_1\,[1 - e^{-k\Delta t_{90}}]}
       {1 - \dfrac{C_1}{C_2} e^{-k\Delta t_{90}}}
$

which reduces to a simple equation in $E = e^{-k\Delta t_{90}}$:

$
E = \frac{0.1}{1 - 0.9\,\dfrac{C_1}{C_2}}
\quad\Rightarrow\quad
t_{90} = -\frac{1}{k(C_1 - C_2)} \ln E
$

```python
import numpy as np
from scipy.optimize import fsolve

# Given parameters from the problem
k = 0.5        # M^-1 s^-1
CA0 = 0.444    # M (inlet concentration after mixing)
CA_ss = 0.020  # M (steady-state outlet concentration required)

# We need tau from Q1.2 (revised)
tau = 1960 / 0.922   # s

# Equation: k*C^2 + (1/tau)*C - (CA0/tau) = 0
def steady_state_roots():
    a = k
    b = 1/tau
    c = -CA0/tau
    roots = np.roots([a, b, c])
    return roots

C1, C2 = steady_state_roots()

# Pick physical root
C1_pos = max(C1, C2)
C2_neg = min(C1, C2)

# Step 2: Compute t90 numerically using the exact analytic solution
Delta = C1_pos - C2_neg

E = 0.1 / (1 - 0.9 * (C1_pos / C2_neg))

t90 = -np.log(E) / (k * Delta)

print(f'Steady state concentration: at C_SS ={C1_pos:.3f} M')
print(f't_90 = {t90:.3f}s = {t90/60:.3f}m')
```

An autocatalytic reaction is one in which a reaction product catalyzes the formation of more product. Many biological processes, for instance, are autocatalytic. You are designing a reactor for the autocatalytic, solution-phase, constant-density reaction:

$\ce{2 A -> B}\ \ \ \ \ \ \ \ r=kC_AC_B\ \ \ \ \ \ \ \ \ k=0.5 \text{M}^{-1}\text{s}^{-1}$

starting from 2.0 M solution of A and producing 1 mol/s of 0.9 M B.

### Q2.1. Determine the reactor volume and residence time necessary to run the reaction in a CSTR.

### 1. Volumetric flow rate

$
\dot V = \frac{F_{B,\text{out}}}{C_{B,\text{out}}}
       = \frac{1\ \text{mol/s}}{0.9\ \text{mol/L}}
       = 1.11\ \text{L/s}
$

---

### 2. Outlet concentrations of A and conversion

Stoichiometry (no B in feed):

$
2A \rightarrow B \quad\Rightarrow\quad
C_B = \frac{C_{A0} - C_A}{2}

Given $C_{A0}=2.0\ \text{M}$ and $C_B = 0.9\ \text{M}$:

$
C_{A0} - C_A = 2 C_B = 1.8
\Rightarrow C_A = 2.0 - 1.8 = 0.2\ \text{M}
$

---

### 3. CSTR design
CSTR design equation for A:
$$\frac{C_{A} - C_{A0}}{\tau}= -2k C_A C_B$$ 
$\tau$ is the only unknown

Rearrange and plug in: $C_{A0}=2.0\ \text{M}, C_A = 0.2\ \text{M}, C_B = 0.9\ \text{M}, k = 0.5 \text{M}^{-1}\text{s}^{-1}$

$\boxed{\tau = \frac{C_{A0} - C_A}{2k C_A C_B} = \frac{0.2\ \text{M} - 2.0\ \text{M}}{2 0.5 \text{M}^{-1}\text{s}^{-1} 0.2\ \text{M} 0.9\ \text{M}} = 10\ \text{s}}$

### 4. CSTR volume
$\tau = \frac{V}{\dot V} \quad\Rightarrow\quad \boxed{V = \tau \dot V = 10\ \text{s} \times 1.11\ \text{L/s} \approx 11.1 \text{L}}$

### Q2.2. Same question, PFR.

### PFR design in terms of conversion

Let conversion of A be $X$:

- Constant-density → volumetric flow rate $\dot V$ is constant.
- Batch reactor design equation:
$$ \frac{dC_B}{d\tau} = k C_A C_B$$

Rearrange and integration from $\tau = 0$ to $\tau$ and $C_B$ to $C_{B0}$


$ \ln\left(\frac{C_B}{C_{B0}}\right) = k C_A \tau $

We have $C_{B0}=0$, so $\boxed{V = \infty, \tau = \infty}$

### Q2.3. Suppose you want to minimize total reactor volume by staging a PFR and CSTR in series. In which order should you put the reactors, and what reactor volumes should you choose?

Desired outlet: $C_B = 0.9\ \text{M}\Rightarrow X_f = 0.90$.  
From Q2.1:
- Volumetric flow: $\dot V = 1.11\ \text{L/s}$
- $F_{A0} = C_{A0}\dot V = 2.22\ \text{mol/s}$

For constant density and no B in the feed,

$
C_A = C_{A0}(1-X), \qquad
C_B = \frac{C_{A0}X}{2}
$

Rate in terms of conversion:

$
-r_A = 2r = k C_{A0}^2 X(1-X)
$
with $k=0.5,\ C_{A0}=2.0 \Rightarrow -r_A = 2 X(1-X)\ \text{mol·L}^{-1}\text{s}^{-1}$

Levenspiel “design function”:

$
f(X) = \frac{F_{A0}}{-r_A} = \frac{F_{A0}}{2X(1-X)}
$

To minimize total volume with a **CSTR + PFR** in series, we choose the
switching conversion $X_1$ where $f(X)$ is minimum (i.e. where
$-r_A$ is maximum):

$
\frac{df}{dX} = 0 \quad\Rightarrow\quad X_1 = 0.5
$

So the optimum arrangement is:

> **CSTR first (0 → 0.5), then PFR (0.5 → 0.9).**

---

### 1. Volume of the CSTR (0 → 0.5)

At $X_1 = 0.5$:

$
-r_A(X_1) = 2(0.5)(0.5) = 0.5\ \text{mol·L}^{-1}\text{s}^{-1}
$

CSTR design:

$
V_1 = \frac{F_{A0}(X_1 - 0)}{-r_A(X_1)}
    = \frac{2.22 \times 0.5}{0.5}
    \approx 2.22\ \text{L}
$

---

### 2. Volume of the PFR (0.5 → 0.9)

PFR design:

$
V_2 = \int_{X_1}^{X_f} \frac{F_{A0}}{-r_A}\,dX
    = \int_{0.5}^{0.9} \frac{2.22}{2X(1-X)}\,dX
    = \frac{2.22}{2}\left[\ln\frac{X}{1-X}\right]_{0.5}^{0.9}
$

$
\frac{X_f}{1-X_f} = \frac{0.9}{0.1} = 9, \quad
\frac{X_1}{1-X_1} = 1
$

$
V_2 = 1.11 \ln 9 \approx 1.11 \times 2.197 \approx 2.44\ \text{L}
$

---

### 3. Total volume and ordering

$
V_{\text{total}} = V_1 + V_2
                 \approx 2.22 + 2.44
                 \approx 4.66\ \text{L}
$

Compare with **single CSTR** from Q2.1: $V \approx 11\ \text{L}$.  
So staging **cuts the volume by more than half**.

# Q3. Trade-offs
A 1 M solution of reactant A is used to make B in an isothermal, constant density reactor. Unfortunately, the reaction also makes an undesired product D. The reaction sequence and kinetics are:

$\ce{2 A -> B}\ \ \ \ \ \ \ r_1=k_1C_A^2$

$\ce{A + B-> D}\ \ \ \ \ \ \ r_2=k_2C_AC_B$

where $k_1 = k_2 = 0.2 \text{M}^{-1}\text{s}^{-1}$.

### Q3.1. Plot the concentrations of all species vs residence time up to 10 min assuming the reaction is run in a PFR.

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

# Kinetics
k1 = 0.2  # M^-1 s^-1
k2 = 0.2  # M^-1 s^-1

# ODE system in terms of residence time tau (s)
# Reactions:
# 2 A -> B     with r1 = k1 * CA^2
# A + B -> D   with r2 = k2 * CA * CB
def dCdtau(tau, C):
    CA, CB, CD = C
    r1 = k1 * CA**2
    r2 = k2 * CA * CB
    
    dCA = -2 * r1 - r2
    dCB = +1 * r1 - r2
    dCD = +1 * r2
    return [dCA, dCB, dCD]

# Initial conditions at reactor inlet (tau = 0)
C0 = [1.0, 0.0, 0.0]   # CA0 = 1 M, CB0 = 0, CD0 = 0

# Integrate up to 10 min = 600 s
tau_span = (0.0, 600.0)
tau_eval = np.linspace(0.0, 600.0, 400)

sol = solve_ivp(dCdtau, tau_span, C0, t_eval=tau_eval)

tau_PFR = sol.t
CA, CB, CD = sol.y

# Plot concentrations vs residence time
plt.figure()
plt.plot(tau_PFR/60, CA, label="C_A")
plt.plot(tau_PFR/60, CB, label="C_B")
plt.plot(tau_PFR/60, CD, label="C_D")
plt.xlabel("Residence time τ (min)")
plt.ylabel("Concentration (M)")
plt.title("Concentrations vs residence time in PFR")
plt.legend()
plt.show()
```

### Q3.2. Plot the concentrations of all species vs residence time up to 10 min assuming the reaction is run in a CSTR.

Using CSTR design equation:
$$\frac{C_{i} - C_{i0}}{\tau}= \nu_i r_i$$ 
Derive the concentration equations for species:

$ C_A = \tau (-2 r_1 - r_2) + C_{A0}$

$ C_B = \tau (r_1 -r_2)$

$ C_D = \tau (r_2)$

where $ r_1 = k_1 C_A^2$ and $ r_2 = k_2 C_A C_B$

solve nonlinear algebraic equations using fsolve from scipy.optimize package

```python
from scipy.optimize import fsolve

CA0 = 1.0

def cstr_equations(vars, tau):
    CA, CB, CD = vars
    r1 = k1 * CA**2
    r2 = k2 * CA * CB

    # Define three concentration equations by moving everything in RHS to LHS and setting 0 = ...
    eq1 = CA - (CA0 + tau*(-2*r1 - r2))
    eq2 = CB - (tau*(r1 - r2))
    eq3 = CD - (tau*(r2))
    return [eq1, eq2, eq3]

tau_CSTR = np.linspace(0, 600, 300)
CA_list, CB_list, CD_list = [], [], []

for tau in tau_CSTR:
    sol = fsolve(cstr_equations, [1.0, 0.0, 0.0], args=(tau,))
    CA_list.append(sol[0])
    CB_list.append(sol[1])
    CD_list.append(sol[2])

plt.figure()
plt.plot(tau_CSTR/60, CA_list, label="C_A")
plt.plot(tau_CSTR/60, CB_list, label="C_B")
plt.plot(tau_CSTR/60, CD_list, label="C_D")
plt.xlabel("Residence time τ (min)")
plt.ylabel("Concentration (M)")
plt.title("CSTR Concentrations vs Residence Time")
plt.legend()
plt.show()
```

### Q3.3. Which reactor would you choose if you want to maximine the *yield* of B?

PFR because $C_B$ rises to a higher peak

### Q3.4. Which reactor would you choose if you want to maximine the *selectivity* torwards B?

```python
plt.figure()
plt.plot(tau_PFR[1:]/60, CB[1:]/CD[1:], label="PFR")
CB_list = np.array(CB_list[1:])
CD_list = np.array(CD_list[1:])
plt.plot(tau_CSTR[1:]/60, CB_list/CD_list, label="CSTR")
plt.legend()
plt.show()
```

Because the CSTR keeps A and B fully mixed at the same composition, B is immediately consumed to D, whereas the PFR separates high-A and high-B regions along the reactor, allowing B to accumulate and giving consistently higher B/D selectivity.
