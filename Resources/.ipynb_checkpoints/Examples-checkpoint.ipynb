## Example - Flow rate calculation

Air is 80% N2 and 20% O2 (mol/mol). A mass flow controller delivers 1 g/s of dry air to a reactor at 298 K and 1 bar. What are the molar flow rates of N2, O2, and the total volumetric flow rate?

```python
T = 298 ; P =1.0e5 # K, Pa
```

```python
yN2 = 0.8; yO2=0.2
```

```python
MWN2 = 28; MWO2 = 32
```

```python
MWair = yN2*MWN2 + yO2*MWO2 # g/mol
```

```python
massflowrate = 1 # g air /s
```

```python
FN2 = massflowrate * MWair * yN2 # mol/s
```

```python
FO2 = massflowrate * MWair * yO2 # mol/S
```

```python
print("N2, O2 molar flow rates = {0:6.2f} and {1:6.2f} mol/s".format(FN2,FO2))
```

```python
Vdotair = (FN2 + FO2) * const.gas_constant * T / P
```

```python
print("Volumetric flow rate = {0:8.3f} m3/s".format(Vdotair))
```

## Example 2 - Integrating rate laws/numerical integration
A + B -> C, rate = k CA CB. Integrate and plot concentrations vs time.

```python
import numpy as np               #this lets up handles arrays of data
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp

def dCdt(C,t,k):
    dC_Adt = -k*C[0]*C[1]     # A + B -> C;  r = k CA CB
    dC_Bdt = -k*C[0]*C[1]
    dC_Cdt =  k*C[0]*C[1]
    dCdt = [dC_Adt,dC_Bdt,dC_Cdt] 
    return dCdt

# initialize concentrations
C_0 = [1., 1.5, 0.]

# initialize k's
k = 0.2

# Range of time to solve over
t = np.arange(0,10,0.1) 

p = (k,) # turn parameters into a tuple
# Solve two ODEs with odeint
C = odeint(dCdt,C_0,t,p)

C_A = C.transpose()[0] # Get C_A from C
C_B = C.transpose()[1] # Get C_B from C
C_C = C.transpose()[2]
plt.figure()
plt.plot(t,C_A,'-',label=r'$C_{\rm A}$')
plt.plot(t,C_B,'-',label=r'$C_{\rm B}$')
plt.plot(t,C_C,'-',label=r'$C_{\rm C}$')
plt.xlabel('Time (s)')
plt.ylabel('Concentration (mol/L)')
plt.legend()
```

## Example 3 - Extracting kinetics from concentration vs time data
m-xylene + Br2 -> Br-m-xylene with an I2 catalyst. Reaction is run at 17 C in xylene solvent and the concentration of Br2 observed vs time. What is the rate order?

### First plot experimental data

```python
import numpy as np               #this lets up handles arrays of data
import matplotlib.pyplot as plt

# Experimental data:
t = np.array([0.00, 2.25, 4.50, 6.33, 8.00, 10.25, 12.00, 13.50, 15.60, 17.85, 19.60, 27.00, 30.00, 38.00, 41.00, 45.00, 47.00, 57.00, 63.00])

C_Br2 = np.array([0.3335, 0.2965, 0.2660, 0.2450, 0.2255, 0.2050, 0.1910, 0.1794, 0.1632, 0.1500, 0.1429, 0.1160, 0.1053, 0.0830, 0.0767, 0.0705, 0.0678, 0.0553, 0.0482])

plt.figure()
plt.plot(t,C_Br2,'o')
plt.xlabel('Time (s)')
plt.ylabel('Concentration (mol/L)')
```

### Test whether reactions is first order in Br2 - it isn't!

```python
# If first order, ln C vs 1/t should be linear. Remember to remove t=0 data point
lnC_Br2 = np.log(C_Br2[1:])
Inv_t = 1/t[1:]
plt.figure()
plt.plot(Inv_t, lnC_Br2,'o')
plt.xlabel('1/Time (s$^{-1}$)')
plt.ylabel('ln(Concentration (mol/L))')
plt.title('Decidedly not a line!')
```

### Use finite difference to extract rates

```python
grad_t = np.gradient(t)            # second order approximation to gradient, allowing for unequal step size
grad_C = np.gradient(C_Br2)
rate = -np.divide(grad_C,grad_t)

plt.figure()
plt.plot(C_Br2,rate,'o')
plt.xlabel('Concentration (mol/L)')
plt.ylabel('Rate (mol/L/s)')
plt.title('Rate of Br2 disappearance vs concentration')
```

### Test whether data fits a power law model

```python
from scipy.optimize import curve_fit

# define rate model with unknown k and alpha
def rate_model(x, k, alpha):
    return k*x**alpha

# attempt to fit rate vs C_Br2 data to differential model
popt, pcov = curve_fit(rate_model, C_Br2, rate )

print('k = {0:f}, alpha={1:f}'.format(popt[0],popt[1]))

model_prediction = rate_model(C_Br2,popt[0],popt[1])

plt.figure()
plt.plot(C_Br2,rate,'o')
plt.plot(C_Br2,model_prediction,'-')
plt.xlabel('Concentration (mol/L)')
plt.ylabel('Rate (mol/L/s)')
plt.title('Observed and modeled rate of Br2 disappearance')
```

### Plot difference between model and fit. Is it random?

```python
difference_array = np.subtract(rate, model_prediction)
squared_array = np.square(difference_array)
mse = squared_array.mean()
print('Mean square error = {}'.format(mse))
plt.figure()
plt.plot(C_Br2,difference_array,'o')
plt.xlabel('Concentration (mol/L)')
plt.ylabel('Model-observed rate (mol/L/s)')
plt.plot([C_Br2[0],C_Br2[-1]],[0,0],'-')
```

### Test whether a 3/2 order model would fit the data. First figure out the integrated rate law

```python
from sympy import Function, Eq, dsolve, Symbol
# 1. Define the symbolic variables
time = Symbol('time'); kconst = Symbol('kconst')
Conc = Function('Conc')

# 2. Define the differential equation
# y(x).diff(x) represents the derivative dy/dx
# The Eq() function sets the left side equal to the right side (0 in this case)
equation = Eq(Conc(time).diff(time), -kconst*Conc(time)**(3/2))

print("Original Equation:")
display(equation)

# 3. Solve the differential equation using dsolve()
solution = dsolve(equation, Conc(time))

print("\nSymbolic Solution:")
display(solution)
print('4/C1**2 must be C0, or C1 = 2/C0**0.5')
```

### Fit data to integral model

```python
def integral(t, C0, k):
    return (4*C0/(2+k*C0**0.5*t))**2

popt1, pcov1 = curve_fit(integral, t, C_Br2)
print('k = {0:f} 1/s 1/M^1/2'.format(popt[1]))

integral_model = integral(t, popt1[0], popt1[1])

plt.figure()
plt.plot(t,C_Br2,'o')
plt.plot(t,integral_model,'-')
plt.xlabel('Time (s)')
plt.ylabel('Concentration (mol/L)')
```

## Example 4 - Arrhenius plots and linear regression
Rate constansts for the first-order decomposition reaction $\ce{2 N2O5 -> N2O4 +O2}$ were obtained at various temperatures.

### Create Arrhenius plot

```python
import numpy as np               #this lets up handles arrays of data
import matplotlib.pyplot as plt
from scipy import stats          # one of several options for doing linear regression

#experimental data
Temperature = np.array([288,298,313,323,338]) # Kelvin
konstant = np.array([1.04e-5,3.38e-5,2.47e-4,7.59e-4,4.87e-3])

# linearize
lnkonstant = np.log(konstant)
Temp_inverse = 1/Temperature

# make Arrhenius plot
plt.figure()
plt.plot(Temp_inverse,lnkonstant,'o')
plt.xlabel('1/T K$^{-1}$')
plt.ylabel('ln k')
```

### Perform linear regression

```python
from scipy import stats          # one of several options for doing linear regression

result = stats.linregress(Temp_inverse, lnkonstant)
                         
print(f"Slope: {result.slope:.4f}")
print(f"Intercept: {result.intercept:.4f}")
print(f"R-squared: {result.rvalue**2:.4f}")
print(f"P-value: {result.pvalue:.4g}") # Use scientific notation for small p-values
print(f"Standard Error: {result.stderr:.4f}")
```

### Check fit

```python
model = result.intercept + result.slope*Temp_inverse
plt.figure()
plt.plot(Temp_inverse,lnkonstant,'o')
plt.plot(Temp_inverse,model,'-')
plt.xlabel('1/T K$^{-1}$')
plt.ylabel('ln k')
```

### Extract parameters

```python
R0 = 8.31441 # J/mol K
prefactor = np.exp(result.intercept)
activation_energy = -R0*result.slope

print(f"A = {prefactor:.4g} 1/s")
print(f"Ea = {activation_energy/1000:.2f} kJ/mol")
```

## Example 5 - Transition State Theory Calculation
Compute the rate constant for the isomerization of vinyl alcohol to acetaldehye:


### Set up a "Molecule" class to contain fundamental properties of each species and to provide statistical thermodynamic provides

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve

kB = 8.61733e-5      # eV /K
h = 4.13566766e-15  # eV s
eVtokJ = 96.485332
autoeV= 27.212      # eV
Nav = 6.022e23      # Avogadro's number 
amutoeVc2 = 931494095.17 # convert amu to eV/c^2
R0 = kB * eVtokJ * 1000.       # gas constant in J/mol K
c = 299792458 # m / s

class Molecule(object):
    def __init__(self, name, mass, energy, rot, vib, helmholtz):
        self.name = name
        self.mass = mass     # g/mol
        self.energy = energy # au
        self.rot = rot       # theta_rot, K
        self.vib = vib       # theta_vib, K
        self.helmholtz = helmholtz

    def Hello(self):
        return "Hello {}".format(self.name)

    def Lambda(self, T):
        m = self.mass * amutoeVc2
        return h/np.sqrt(2*np.pi*m*kB*T)*c

    def qtrans(self,T):  #  mole/m^3
        m = self.mass * amutoeVc2
        l = h/np.sqrt(2*np.pi*m*kB*T)*c
        return (1./Nav)*(1./l)**3

    def qrot(self,T):
        if self.rot[1] > 0:
            return np.sqrt((np.pi*(T/self.rot[0])*(T/self.rot[1])*(T/self.rot[2])))
        else:
            return T/self.rot[0]

    def qvib(self,T):
        qvib = 1
        for mode in self.vib:
            qvib = qvib*(1.-np.exp(-mode/T))**-1
        return qvib

    def qtot(self,T):
        return self.qtrans(T)*self.qrot(T)*self.qvib(T)

    def ZPE(self): # eV
        ZPE=0.
        for mode in self.vib:
            ZPE = ZPE + mode
        return ZPE * (0.5)*kB

    def Utrans(self,T): # eV
        return 1.5*kB*T

    def Urot(self,T): # eV
        return 1.5*kB*T

    def Uvib(self,T): #eV
        Uvib=0.
        for mode in self.vib:
            Uvib=Uvib+kB*(mode/(np.exp(mode/T)-1.0))
        return Uvib

    def Utot(self,T): #eV
        return self.Utrans(T)+self.Urot(T)+self.Uvib(T)
```

### Define fundamental thermodynamic relationships

```python
def Helmholtz(q):  # Helmholtz energy from partition function
   return -kB*T*np.log(q) #eV

def S(A,U,T): 
   return (U - A)/T

def Delta(nu,A): # compute Delta for an arbitrary reaction
    return np.dot(nu,A)

def stdconc(T): 
    (1e5/(R0*T)) # mol/m3 equivalent of 1 bar at T 
```

### Specify properties of molecules of interest

```python
VA = Molecule("Vinyl alcohol", 44., -153.802227784, [3.06869, 0.49808, 0.42852],[342.12,   692.39,  1032.15,  1198.36,  1386.62, 1411.99,  1670.11,  1880.98,  1967.27,  2101.01, 2532.91,  4541.20,  4591.62,  4716.63,  5454.14],[])

Ac = Molecule("Acetaldehyde", 44., -153.830119460, [2.72923, 0.48351, 0.43371], [222.94,  729.92,  1121.87,  1285.63,  1636.17, 1646.79,  2018.97,  2080.76, 2143.85,  2157.44,  2651.18,  4169.12,  4379.22,  4459.71,  4553.14],[])

TS = Molecule("TS", 44., -153.714172988, [2.33148, 0.58652, 0.48832], [857.94,   939.84,  1148.55,  1395.88,  1536.40, 1678.69,  1744.34,  1877.98,  2155.56,  2261.86, 2665.34,  4471.92,  4522.04,  4596.48], [])

T = np.linspace(600,1500,500)
```

### Overall reaction thermodynamics

```python
deltaE = (Ac.energy - VA.energy)*autoeV
deltaU = deltaE+(Ac.Utot(T) - VA.Utot(T))
deltaH = deltaU  # number of molecules is conserved

deltaA = deltaE+Helmholtz(Ac.qtot(T)) - Helmholtz(VA.qtot(T))
deltaS = S(deltaA,deltaU,T)

plt.figure()
plt.plot(T,(deltaU)*eVtokJ,label="Internal")
plt.plot(T,(deltaA)*eVtokJ,label="Free")
plt.title("VA to Ac reaction energies")
plt.xlabel("Temperature (K)")
plt.ylabel("Reaction Energies (kJ/mol)")
plt.legend()

plt.figure()
plt.plot(T,deltaS*eVtokJ*1000,label="Entropy")
plt.title("VA to AC reaction entropy")
plt.xlabel("Temperature (K)")
plt.ylabel("Reaction Entropy (J/mol/K)")

plt.figure()
plt.plot(1/T,-deltaA/(kB*T))
plt.title("Equilibrium constant vs T")
plt.xlabel("1/Temperature (1/K)")
plt.ylabel("log(Keq)")
```

### Compute (zero-point corrected) activation energy

```python
Ea0 = (TS.energy-VA.energy)*autoeV + (TS.ZPE() - VA.ZPE())  

print('Vinyl alcohol to TS {:4.0f} kJ/mol'.format(Ea0*eVtokJ))
```

### Compute and plot pre-factor

```python
kTh = (kB/h) * T 
qratio = (TS.qtot(T)/VA.qtot(T))
A  = kTh * qratio

plt.figure()
plt.plot(T,A)
plt.xlabel("Temperature (K)")
plt.title("A vs T")
plt.ylabel("A (/s)")
```

### Compute and plot rate constant

```python
k = A * np.exp(-Ea0/(kB*T))

plt.figure()
plt.plot(T,k)
plt.xlabel("Temperature (K)")
plt.title("k vs T")
plt.ylabel("k (/s)")

plt.figure()
plt.plot(1/T,np.log(k))
plt.xlabel("1/Temperature (1/K)")
plt.title("Arrhenius plot: log k vs 1/T")
plt.ylabel("log(k) (/s)")
```

### Plot half-life

```python
thalf = np.log(2)/k

plt.figure()
plt.title("Half life")
plt.plot(T,thalf,label="half life")
plt.xlabel("Temperature (K)")
plt.ylabel("s")
```

### The thermodynamic perspective

```python
deltaUd = Ea0+(TS.Utot(T) - VA.Utot(T))
deltaHd = deltaUd 

deltaAd = Ea0+Helmholtz(TS.qtot(T)) - Helmholtz(VA.qtot(T))
deltaSd = S(deltaAd,deltaUd,T)

plt.figure()
plt.plot(T,(deltaUd)*eVtokJ,label="Internal")
plt.plot(T,(deltaAd)*eVtokJ,label="Free")
plt.title("VA activation energies")
plt.xlabel("Temperature (K)")
plt.ylabel("Activation Energies (kJ/mol)")
plt.legend()

plt.figure()
plt.plot(T,deltaSd*eVtokJ*1000,label="Entropy")
plt.title("VA activation entropy")
plt.xlabel("Temperature (K)")
plt.ylabel("Activation Entropy (J/mol/K)")
```

## Example 7: $\ce{A <-> B <-> C}$ reaction network

### Set up reaction rate models and free energy surface

```python
# A <-> B <-> C reaction network
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp
from scipy.interpolate import BPoly

kB = 8.61733e-5      # eV /K
h = 4.13566766e-15  # eV s
R0 = 8.31441 # J/mol/K
    
# function to plot free energy surface. Takes G_i = [G_TS,G_rxn]
def free_energy_surface(G1,G2):
    p1 = BPoly.from_derivatives([0, 1, 2, 3, 4], [[0, 0], [G1[0], 0], [G1[1], 0], [G1[1]+G2[0],0], [G1[1]+G2[1],0]])
    x=np.linspace(0,4,100)
    y=p1(x)

    plt.figure(figsize=(10,12))
    plt.subplot(2,2,1)
    plt.plot(x,y)
    plt.ylabel('Free energy')
    plt.xlabel('Reaction coordinate')
    plt.title('Reaction Free Energy Landscape')

G1=[125,-75]  # Gddagger, G reaction
G2=[125,-75]

free_energy_surface(G1,G2)
```

```python
def rate_constants(T,G1,G2):
    G1dd = G1[0]
    Gn1dd = G1dd-G1[1]
    G2dd = G2[0]
    Gn2dd = G2dd-G2[1]

    k = (kB*T/h)*np.exp(-np.array([G1dd,Gn1dd,G2dd,Gn2dd])*1000./(R0*T))
    return k

T = 500.
k=rate_constants(T,G1,G2)

print('T = {:4.1f}'.format(T))
print('delta G1dagger = {:4.1f}  delta Gn1dagger = {:4.1f}  delta G2dagger = {:4.1f}   delta Gn2dagger = {:4.1f} kJ/mol'.format(G1[0],G1[0]-G1[1],G2[0],G2[0]-G2[1]))
print('k1 = {:6.3e}     k-1 = {:6.3e}         k2 = {:6.3e}      k-2 = {:6.3e} /s'.format(k[0],k[1],k[2],k[3]))
```

### Integrate for concentrations vs time

```python
# define reaction and species rates
def r1(C,k):
    r1 = k[0]*C[0] - k[1]*C[1]
    return r1

def r2(C,k):
    r2 = k[2]*C[1] - k[3]*C[2]
    return r2

def dCdt(C,t,k): # returns species rates
    dC_Adt = -r1(C,k)
    dC_Bdt =  r1(C,k) - r2(C,k)
    dC_Cdt =  r2(C,k)
    dCdt = [dC_Adt,dC_Bdt,dC_Cdt] 
    return dCdt

def Ct(C_0,k,t): # integrate numerically
    p = (k,) # turn parameters into a tuple
    Concentrations = odeint(dCdt,C_0,t,p)
    C_A = Concentrations.transpose()[0] # Get C_A from C
    C_B = Concentrations.transpose()[1] # Get C_B from C
    C_C = Concentrations.transpose()[2]
    return C_A,C_B,C_C
```

```python
# integrate
C_0 = [1., 0, 0.]

# Range of time to solve over
t = np.arange(0,10,0.1) 

C_A,C_B,C_C=Ct(C_0,k,t)
    
plt.figure()
plt.plot(t,C_A,'-',label=r'$C_{\rm A}$')
plt.plot(t,C_B,'-',label=r'$C_{\rm B}$')
plt.plot(t,C_C,'-',label=r'$C_{\rm C}$')
plt.xlabel('Time (s)')
plt.ylabel('Concentration (mol/L)')
plt.title('Concentration vs time')
plt.legend()
```

### Analytical solution in irreversible limit

```python
def C_irreversible(CA0,k,t):
    C_A=CA0*np.exp(-k[0]*t)
    C_B=CA0*(k[0]/(k[1]-k[0]))*(np.exp(-k[0]*t)-np.exp(-k[1]*t))
    C_C=CA0-C_A-C_B
    Concentrations=np.array([C_A,C_B,C_C])
    return Concentrations

G1=[125,-75]  # Gddagger, G reaction
G2=[128,-75]
k=rate_constants(T,G1,G2)

k_irreversible=[k[0],k[2]]
Concentrations=C_irreversible(C_0[0],[k[0],k[2]],t)

C_A = Concentrations[0] # Get C_A from C
C_B = Concentrations[1] # Get C_B from C
C_C = Concentrations[2]

plt.figure()
plt.plot(t,C_A,'-',label=r'$C_{\rm A}$')
plt.plot(t,C_B,'-',label=r'$C_{\rm B}$')
plt.plot(t,C_C,'-',label=r'$C_{\rm C}$')
plt.xlabel('Time (s)')
plt.ylabel('Concentration (mol/L)')
plt.title('Analytical result for irreversible A->B->C reaction')
plt.legend()
```

### Gillespie algorithm

```python
def gillespie_abc(k, N, C_A0, C_B0, C_C0, t_max):
    """
    Executes the Gillespie Algorithm for the A -> B -> C reaction network.

    Args:
        k1 (float): Rate constant for A -> B.
        k2 (float): Rate constant for B -> C.
        N (int): total number of particles
        C_A0 (int): Initial concentration of molecules of A.
        C_B0 (int): Initial concentration of molecules of B.
        C_C0 (int): Initial concentration of molecules of C.
        t_max (float): Maximum simulation time.

    Returns:
        tuple: (time_points, N_A_history, N_B_history, N_C_history)
    """
    
    # 1. Initialize State and Time
    t = 0.0

    C0 = C_A0 + C_B0 + C_C0     # mol/dm3    \
    N_A = int(N*C_A0/C0)
    N_B = int(N*C_B0/C0)
    N_C = int(N*C_C0/C0)
    
    N_A = N - N_B - N_C
    
    C_scale = C0/N
    
    # Initialize history lists
    time_points = [t]
    C_A_history = [N_A*C_scale]  
    C_B_history = [N_B*C_scale]
    C_C_history = [N_C*C_scale]

    # 2. Main Loop
    while t < t_max and N_A + N_B > 0:
        r1 = k[0] * N_A
        r2 = k[1] * N_B

        rtot = r1 + r2

        # --- b. Calculate Time Step (tau) ---
        random1 = np.random.rand()
        
        # Calculate time until the next reaction (tau)
        # Formula: tau = (1/a0) * ln(1/r1)
        tau = -np.log(random1)/rtot

        # --- c. Determine Which Reaction Occurs (j) ---
        random2 = np.random.rand()
        
        if random2 * rtot < r1:    # reaction 1 happens
            N_A -= 1
            N_B += 1
        elif random2 * rtot < rtot:  # reaction 2 happens
            N_B -= 1
            N_C += 1
        
        t += tau     # advance time
        time_points.append(t)
        C_A_history.append(N_A*C_scale)
        C_B_history.append(N_B*C_scale)
        C_C_history.append(N_C*C_scale)

    return time_points, C_A_history, C_B_history, C_C_history

# --- Simulation Parameters and Execution ---

G1=[125,-75]  # Gddagger, G reaction
G2=[125,-75]
T = 500.

k=rate_constants(T,G1,G2)
k_trim = np.array([k[0],k[2]])

CA0 = 1.0  # Starting concentration of A (M)
CB0 = CC0 = 0.
t_max = 100.0  # Total simulation time
N = 200

# Run the simulation

time, CA, CB, CC = gillespie_abc(k_trim, N, CA0, CB0, CC0, t_max)

# --- Plotting the Results ---
plt.figure()
plt.plot(time, CA, 'o', label='A', color='red', markerfacecolor='none')
plt.plot(time, CB, 'o', label='B', color='blue', markerfacecolor='none')
plt.plot(time, CC, 'o', label='C', color='green', markerfacecolor='none')

plt.title('Gillespie Simulation of A → B → C')
plt.xlabel('Time (s)')
plt.ylabel('Concentrations (M)')
plt.legend()
```

### Sabatier principle

```python
import numpy as np
import matplotlib.pyplot as plt

R = 8.314        # Gas constant (J/mol/K)
T = 300.0        # Temperature (K)
RT = R * T / 1000  # RT in kJ/mol

# --- BEP Parameters (Brønsted-Evans-Polanyi) ---
# These parameters define the linear relationship: Delta G_activation = alpha * Delta G_reaction + beta
# Simplified for a two-step cycle where Delta G_ads is the descriptor (x-axis)
# The slope is 'alpha', the intercept is 'beta'.

# Step 1: Reactant Activation (e.g., A + * -> A*)
# Barrier increases as A* binds weaker (Delta G_ads increases)
ALPHA1 = 0.5     # Slope (change in barrier per change in binding energy)
BETA1 = 50.0     # Baseline activation energy (kJ/mol)

# Step 2: Product Desorption (e.g., A* -> P + *)
# Barrier decreases as A* binds stronger (Delta G_ads decreases)
ALPHA2 = 0.5     # Slope (magnitude is arbitrary but must be opposite of ALPHA1)
BETA2 = 50.0     # Baseline activation energy (kJ/mol)


def calculate_rate(delta_g_ads):
    """Calculates the overall rate constant based on the BEP principle."""

    # 1. Activation Energy for Step 1 (Weak Binding Limit)
    # The barrier for forming the intermediate. Increases as Delta G_ads increases.
    delta_g_1_double_dagger = ALPHA1 * delta_g_ads + BETA1

    # 2. Activation Energy for Step 2 (Strong Binding Limit)
    # The barrier for consuming the intermediate. Decreases as Delta G_ads increases.
    # Note the -(delta_g_ads) term to create the opposite slope for the volcano.
    delta_g_2_double_dagger = ALPHA2 * (-delta_g_ads) + BETA2
    
    # 3. Overall Rate (Limited by the highest barrier - the rate-determining step)
    # The rate is proportional to the Arrhenius factor of the highest barrier.
    # We take the maximum of the two activation energies to find the RDS barrier.
    delta_g_max_double_dagger = np.maximum(delta_g_1_double_dagger, delta_g_2_double_dagger)

    # Rate Constant (simplified Arrhenius expression, assuming pre-factor A=1)
    rate_constant = np.exp(-delta_g_max_double_dagger / RT)
    
    # Return all components for plotting
    return delta_g_1_double_dagger, delta_g_2_double_dagger, rate_constant


# --- Generate Data and Plot ---
# Range of binding energies (x-axis)
DH_ADS = np.linspace(-50, 50, 200) # (kJ/mol)

# Calculate rate components across the range
DG1_BARRIER, DG2_BARRIER, RATE = calculate_rate(DH_ADS)
LOG_RATE = np.log10(RATE) # Use log10 for plot readability

# --- Find the Optimal Point (Volcano Peak) ---
optimal_index = np.argmax(RATE)
optimal_dh = DH_ADS[optimal_index]
optimal_log_rate = LOG_RATE[optimal_index]

# --- Plotting ---
plt.figure(figsize=(9, 6))

# 1. Plot the two BEP lines (Activation Energies)
plt.subplot(2, 1, 1)
plt.plot(DH_ADS, DG1_BARRIER, 'r--', label=r'$\Delta G_1^\ddagger$ (Activation/Weak-Binding Limit)')
plt.plot(DH_ADS, DG2_BARRIER, 'b--', label=r'$\Delta G_2^\ddagger$ (Desorption/Strong-Binding Limit)')
plt.plot(DH_ADS, np.maximum(DG1_BARRIER, DG2_BARRIER), 'k-', linewidth=3, label=r'$\Delta G_{\text{max}}^\ddagger$ (Rate-Determining Step)')
plt.axvline(optimal_dh, color='gray', linestyle=':', alpha=0.7)
plt.title(r'Brønsted–Evans–Polanyi (BEP) Relationship', fontsize=14)
plt.xlabel(r'Intermediate Binding Energy, $\Delta G_{\text{ads}}$ (kJ/mol)')
plt.ylabel(r'Activation Energy, $\Delta G^\ddagger$ (kJ/mol)')
plt.legend(loc='upper left', fontsize=9)
plt.grid(True, which='both', ls='--', alpha=0.6)

# 2. Plot the Volcano Plot (Rate)
plt.subplot(2, 1, 2)
plt.plot(DH_ADS, LOG_RATE, 'g-', linewidth=3)
plt.axvline(optimal_dh, color='r', linestyle='--', label=r'$\Delta G_{\text{ads, opt}}$')
plt.title('Volcano Plot (Sabatier Principle)', fontsize=14)
plt.xlabel(r'Intermediate Binding Energy, $\Delta G_{\text{ads}}$ (kJ/mol)')
plt.ylabel(r'$\log_{10}(\text{Rate})$ (Log Activity)')
plt.text(optimal_dh + 2, optimal_log_rate - 0.5, f'Optimal: {optimal_dh:.1f} kJ/mol', color='r', fontsize=10)
plt.grid(True, which='both', ls='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

### Semi-batch
Flow A into a reactor of B

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp

def Volume(t):
    return V0+v*t

def rate(k,N,V):
    return k*N[0]*N[0]*N[1]/V**2

def dNsemidt(N,t,F0,k): # returns species rates
    V=Volume(t)
    dN_Adt = F0[0] - rate(k,N,V)
    dN_Bdt = -rate(k,N,V) 
    dN_Edt = rate(k,N,V)
    dNdt = [dN_Adt,dN_Bdt,dN_Edt] 
    return dNdt

V0 = 378  # liter
v = 3.78  # liter/min
# Range of time to solve over
t = np.arange(0,5000,5) 
Vol = Volume(t)

plt.figure()
plt.plot(t, Vol)
plt.xlabel('time (min)')
plt.ylabel('volume (l)')
plt.title('Volume vs time')
```

## Non-steady state CSTR
A -> B, rate = kC_A

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp

def rate(k,C):
    return k*C[0]

def dCnonsteady(C,t,C0,tau,k): # returns species rates
    dC_Adt = (C0[0]-C[0])/tau - rate(k,C)
    dC_Bdt = -C[1]/tau + rate(k,C) 
    dCdt = [dC_Adt,dC_Bdt] 
    return dCdt

t = np.arange(0,5,.1) 

k = 2
tau = 1
C0 = [1,0]   # only A in feed

C00 = [0,0]  # nothing in reactor
p = (C0,tau,k) # turn parameters into a tuple
Cj = odeint(dCnonsteady,C00,t,p)

C_A = Cj.transpose()[0]
C_B = Cj.transpose()[1]

plt.figure()
plt.plot(t, C_A,label='C_A')
plt.plot(t, C_B,label='C_B')
plt.legend()
plt.xlabel('time (min)')
plt.ylabel('Concentration (M)')
plt.title('Concentration vs time')
```

## PFR vs CSTR
compare residence times and volumes necessary to achieve a conversion
NO oxidation reaction

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp, quad

def rate(k,C):
    return k*C[0]*C[0]*C[1]

def dCPFR(C,t,C0,tau,k):
    r = rate(k,C)
    dC_NOdt = C0[0] + nu[0]*r
    dC_O2dt = C0[1] + nu[1]*r 
    dC_NO2dt = C0[2] + nu[2]*r
    dCdt = [dC_NOdt,dC_O2dt,dC_NO2] 
    return dCdt

# Universal
Rl = 0.082057          # l atm/mol K

# reaction-dependent
k = 0.7e6             # M^-2 s^-1
nu = np.array([-2,-1,2])

# conditions
T = 298.              # K
P = 1.                # atm
yO2 = 0.1
yNO = 100e-6
yNO2 = 0.
v = 10.               # l/s

y0 = np.array([yNO,yO2,yNO2])   # NO, O2, NO2
C0 = (P/(Rl*T))*y0

# CSTR
conversion = 0.9      # NO limiting
xi = conversion*C0[0]/(-nu[0])
C = C0 + nu*xi

tauCSTR =(C[0]-C0[0])/(nu[0]*rate(k,C))
vCSTR = v * tauCSTR

# PFR
def rateinv(C):
    return 1./(nu[0]*k*C*C*C0[1])
    
tauPFR,error = quad(rateinv,C0[0],C[0])
vPFR = v * tauPFR
print(f"tau_CSTR: {tauCSTR:.0f} s   tau_PFR: {tauPFR:.0f} s")
print(f"V_CSTR:  {vCSTR:.0f} l   V_PFR: {vPFR:.0f} l")
```

## Internal mass transfer
Spherical particle, first-order kinetics

```python
import numpy as np
import matplotlib.pyplot as plt

def Psi(lamb,Phi):
    return np.sinh(lamb*Phi)/(lamb*np.sinh(Phi))

def eta_sphere(Phi):
    return (3./Phi)*(1/np.tanh(Phi) - 1/Phi)

def eta_general(Phi):
    return np.tanh(Phi)/Phi
                              
lamb = np.arange(0.01,1,0.01)

plt.figure()
Phi = 10
plt.plot(lamb,Psi(lamb,.5),label='Thiele = 0.5')
plt.plot(lamb,Psi(lamb,1.5),label='1.5')
plt.plot(lamb,Psi(lamb,3),label='2.5')
plt.plot(lamb,Psi(lamb,4),label='3.5')
plt.plot(lamb,Psi(lamb,5),label='4.5')
plt.ylim(0,1)
plt.legend()
plt.xlabel('lambda')
plt.ylabel('Psi=CA/CAsurf')
plt.title('Concentration profile for 1st order reaction with diffusion')
```

```python
thiele = np.logspace(-1,2,100)
plt.xscale('log')
plt.yscale('log')
plt.plot(thiele,eta_sphere(thiele),label='sphere')
plt.plot(thiele,eta_general(thiele),label='general')
plt.xlabel('log Thiele modulus')
plt.ylabel('log effectiveness')
plt.title('Effectiveness factor for 1st order reaction')
plt.legend()
```

## Second order
Boundary value problem

```python
from scipy.integrate import solve_bvp
import numpy as np

def fun(x, y):
    # y is a vector [y, y']
    Phi = 3
    dpsi_dl = y[1]
    d2psi_dl2 = (2/3) * (Phi*y[0])**2  - (2/x) * y[1]
    return np.vstack([dpsi_dl,d2psi_dl2]) 

# 2. Define the Boundary Conditions
def bc(ya, yb):
    # ya[0,1] left boundary condition: dpsi_dl(0) = 0
    # yb[0,1] right boundary condition: psi(1)=1
    return np.array([ya[1] , yb[0] - 1])

# 3. Solve (requires an initial guess mesh and solution)
x_mesh = np.linspace(0.1, 1, 10) # Initial mesh points
y_guess = np.zeros((2, x_mesh.size)) # Initial guess: y=0, y'=0

result = solve_bvp(fun, bc, x_mesh, y_guess)

plt.plot(result.x,result.y[0],label='2nd order')
plt.plot(result.x,Psi(result.x,3),label='1st order')
plt.legend()
plt.xlabel('lambda')
plt.ylabel('Psi=CA/CAsurf')
plt.title('1st vs 2nd order concentration profile')
```

## Internal plus external mass transfer
Spherical particle, first order kinetics

```python
def Psix(lamb,Phi,B):
    sinhphi = np.sinh(Phi)
    return np.sinh(lamb*Phi)/(lamb*sinhphi+(lamb/B)*(Phi*np.cosh(Phi)-sinhphi))

lamb = np.arange(0.01,1,0.01)

plt.figure()

plt.plot(lamb,Psix(lamb,3,1),label='B = 1')
plt.plot(lamb,Psix(lamb,3,10),label='B = 10')
plt.plot(lamb,Psix(lamb,3,100),label='B = 100')

plt.ylim(0,1)
plt.legend()
plt.xlabel('lambda')
plt.ylabel('Psi=CA/CAsurf')
plt.title('Concentration profile for 1st order reaction with internal and external resistance')
```

### Isothermal PFR
A -> B reaction, rate = k CA, exothermic

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint, solve_ivp, simpson, quad

R0 = 8.31441 # J/mol K

def CA(X):
    return CA0*(1-X)

def rate_iso(X,k):
    return k*CA(X)

def dXdtau_isothermal(X,t,k0):
    dXdtau = [k0*(1-X[0])]
    return dXdtau

V      = 10      # liter
CA0    = 2.0     # M
k300   = 0.05    # /min
deltaH = -90000  # J/mol
max_conversion = 0.9

tau = -np.log(1-max_conversion)/k300
v0  = V/tau

t = np.arange(0,tau,.1) 
X0 = [0]

p=(k300,)

Conversion = odeint(dXdtau_isothermal,X0,t,p)

Xiso = Conversion.transpose()[0]
Volume = t * v0

plt.figure()
plt.plot(t,Xiso)
plt.xlabel('Space time (min)')
plt.ylabel('Conversion')
plt.title('Isothermal conversion vs reactor volume')

rate_isothermal = rate_iso(Xiso,k300)
plt.figure()
plt.plot(Xiso,1./rate_isothermal)
plt.title('Isothermal Levenspiel plot')
plt.xlabel('Conversion')
plt.ylabel('1/rate')

Qdot = -rate_isothermal*deltaH
plt.figure()
plt.plot(t,Qdot)
plt.xlabel('Space (time)')
plt.ylabel('Isothermal heat removal (W/liter)')
      
```

```python
total_heat_removal = simpson(Qdot,Volume)
print(f"Total heat removal = {total_heat_removal/60:.0f} W")
```

### Adiabatic PFR
Same A-> B reaction

```python
def T_adiabatic(X):
    return T0 - (CA0 * deltaH/rhoCp)*X

def k(T):
    A = 2.9e20 # /min
    Ea = 125000 # J/mol
    k = A*np.exp(-125000/(R0*T))
    return k

def dXdtau_adiabatic(X,t):
    Temperature = T_adiabatic(X[0])
    k0 = k(Temperature)
    dXdtau = [k0*(1-X[0])]
    return dXdtau

def rate_adiabatic(X):
    Temperature = T_adiabatic(X)
    k0 = k(Temperature)
    rate_adi = k0*CA0*(1-X)
    return rate_adi

def inv_rate_adiabatic(X):
    Temperature = T_adiabatic(X)
    k0 = k(Temperature)
    inv_rate_adi = 1./(k0*(1-X))
    return inv_rate_adi

rhoCp  = 4000    # J/l K
T0     = 300     # K

X = np.arange(0,1,0.05)

plt.figure()
plt.plot(T_adiabatic(X),X,label='adiabatic')
plt.plot([T0,T0],[0,1],label='isothermal')
plt.xlabel('Temperature (K)')
plt.ylabel('Conversion')
plt.legend()

X0 = [0]
p = ()
t = np.arange(0,tau/10,.01) 
Conversion = odeint(dXdtau_adiabatic,X0,t,p)

Xadi = Conversion.transpose()[0]
Volume = t * v0

plt.figure()
plt.plot(t,Xadi)
plt.xlabel('Space time (min)')
plt.ylabel('Conversion')
plt.title('Adiabatic conversion vs space time')

Xl = np.arange(0,0.9999,0.01)
plt.figure()
plt.plot(Xl,1./rate_adiabatic(Xl))
plt.title('Adiabatic Levenspiel plot')
plt.xlabel('Conversion')
plt.ylabel('1/rate')

tau90=quad(inv_rate_adiabatic,0,0.9)
print(f"Adiabatic reactor volume to 90% conversion = {tau90[0]*v0:.2f} l")
print(f"Adiabatic reactor space time to 90% conversion = {tau90[0]:.1f} min")
```

### PFR with heat exchange

```python
def rate_hx(X,T): # first order
    k0 = k(T)
    rate_hx= k0*CA0*(1-X)
    return rate_hx

def dfdtau(f,t,Ua): # f is a vector of X and T
    rate = rate_hx(f[0],f[1])
    dXdtau = rate_hx(f[0],f[1])/CA0
    dTdtau = (-deltaH * rate - Ua *(f[1]-Text))/rhoCp
    dfdtau = [dXdtau,dTdtau] 
    return dfdtau

Text = 295  # K
T0 = 300   # K
CA0 = 2.0  # M
tau = 10

t = np.arange(0,tau,.01) 

plt.figure()
for Ua in range(0,2000,250):  # W/K
    p = (Ua,)
    f0=(0,T0)
    state = odeint(dfdtau,f0,t,p)
    Xhx = state.transpose()[0]
    Thx = state.transpose()[1]
    plt.plot(Thx,Xhx,label=Ua)

plt.legend()
plt.title("Temperature trajectories with increasing heat exchange")
plt.xlabel("Temperature (K)")
plt.ylabel("Conversion")

t = np.arange(0,tau/3,.01) 
plt.figure()
for Ua in (0,800):  # W/K
    p = (Ua,)
    f0=(0,T0)
    state = odeint(dfdtau,f0,t,p)
    Xhx = state.transpose()[0]
    Thx = state.transpose()[1]
    plt.plot(t,Xhx,label=Ua)

plt.legend()
plt.title("Conversion, Adiabatic vs Heat Exchange")
plt.xlabel("Space time (min)")
plt.ylabel("Conversion")

t = np.arange(0,tau/3,.01) 
plt.figure()
for Ua in (0,800):  # W/K
    p = (Ua,)
    f0=(0,T0)
    state = odeint(dfdtau,f0,t,p)
    Xhx = state.transpose()[0]
    Thx = state.transpose()[1]
    plt.plot(t,Thx,label=Ua)

plt.legend()
plt.title("Temperature, Adiabatic vs Heat Exchange")
plt.xlabel("Space time (min)")
plt.ylabel("Temperature (K)")

# Levenspiel plot
plt.figure()
t = np.arange(0,tau/3,.01) 

for Ua in (0,800):  # W/K
    p = (Ua,)
    f0=(0,T0)
    state = odeint(dfdtau,f0,t,p)
    Xhx = state.transpose()[0]
    Thx = state.transpose()[1]
    rate_inv = 1./rate_hx(Xhx,Thx)
    plt.plot(Xhx,rate_inv,label=Ua)

plt.ylim(0,10)
plt.legend()
plt.title("Levenspiel Plot with Heat Exchange")
plt.ylabel("1/rate")
plt.xlabel("Conversion")
```

### Adiabatic reversible reaction
2 A -> B

```python
from scipy import optimize

def T_adiabatic(X):
    return T0 - (CA0 * deltaH/rhoCp)*X

def k(T):
    Af = 1e10 # /s M
    Ar = 1e14 # /s
    kf = Af*np.exp(-Eaf/(R0*T))
    kr = Ar*np.exp(-Ear/(R0*T))
    k=[kf,kr]
    return k

def rate_rev(X):
    T = T_adiabatic(X)
    k0 = k(T)
    rate_rev= k0[0]*(CA0*(1-X))**2 - k0[1]*CA0*X/2
    return rate_rev

def dXdtau_rev(X,t): 
    rate = rate_rev(X[0])
    dXdtau = rate/CA0
    return [dXdtau]

def equilibrium(X,T):
    k0 = k(T)
    equilibrium = k0[0]*(CA0*(1-X))**2 - k0[1]*CA0*X/2
    return equilibrium

rhoCp  = 4000    # J/l K
Eaf = 90000
deltaH = -40000
Ear = Eaf-deltaH
CA0 = 8

# Equilibrium conversion
Teq = []
Xeq = []
for T in range(250,1000,10):
    sol = optimize.root_scalar(equilibrium, bracket=[0, 1], args=(T,), method='brentq')
    Teq.append(T)
    Xeq.append(sol.root)

plt.figure()
plt.plot(Teq,Xeq)
plt.xlabel('Temperature (K)')
plt.ylabel('X equilibrium')
plt.title('Equilbrium conversion, 2A -> B')

plt.figure()

# Coversion vs time
X0 = [0]
p = ()
t = np.arange(0,tau,1) 

plt.figure()
for T0 in range(340,440,20):  # W/K
    Conversion = odeint(dXdtau_rev,X0,t,p)
    Xrev = Conversion.transpose()[0]
    plt.plot(T_adiabatic(Xrev),Xrev,label=T0)

plt.xlabel('Temperature (K)')
plt.ylabel('Conversion')
plt.title('Reversible adiabatic temperature trajectory')
plt.legend()


plt.figure()
for T0 in range(340,440,20):  # W/K
    Conversion = odeint(dXdtau_rev,X0,t,p)
    Xrev = Conversion.transpose()[0]
    plt.plot(t,Xrev,label=T0)

plt.xlabel('Space time (s)')
plt.ylabel('Conversion')
plt.title('Conversion vs space time, reversible reaction')
plt.legend()
```

### Adiabatic CSTR

```python
from scipy import optimize

def T_adiabatic(X):
    return T0 - (CA0 * deltaH/rhoCp)*X

def k(T):
    A = 2.9e20 # /min
    Ea = 125000 # J/mol
    k = A*np.exp(-125000/(R0*T))
    return k

def rate_adiabatic(X):
    Temperature = T_adiabatic(X)
    k0 = k(Temperature)
    rate_adi = k0*CA0*(1-X)
    return rate_adi

def tau_adiabatic(X):
    tau = CA0 * X / rate_adiabatic(X)
    return tau

def tau_root(X,tau):
    diff = tau_adiabatic(X) - tau
    return diff
    
CA0 = 2 #M
rhoCp  = 4000    # J/l K
T0     = 300     # K
deltaH = -90000  # J/mol

Xl = np.arange(0,0.9999,0.01)
tau = tau_adiabatic(Xl)

plt.figure()
plt.plot(tau,Xl)
plt.xlabel('Space time (min)')
plt.ylabel('Conversion')
plt.title('Adiabatic CSTR')
plt.plot([0.9,0.9],[0,1])

sol_low = optimize.root_scalar(tau_root, bracket=[0, 0.1], args=(0.9,), method='brentq')
sol_med = optimize.root_scalar(tau_root, bracket=[0.2, 0.4], args=(0.9,), method='brentq')
sol_hig = optimize.root_scalar(tau_root, bracket=[0.8, 0.99], args=(0.9,), method='brentq')

rate_low = rate_adiabatic(sol_low.root)
rate_med = rate_adiabatic(sol_med.root)
rate_hig = rate_adiabatic(sol_hig.root)
plt.scatter([0.9,0.9,0.9],[sol_low.root,sol_med.root,sol_hig.root])
print(f"Conversions at tau = 0.9: {sol_low.root:.2f}, {sol_med.root:.2f}, {sol_hig.root:.2f}") 

plt.figure()
plt.plot(Xl,1/rate_adiabatic(Xl))
plt.xlabel('Conversion')
plt.ylabel('1/rate')
plt.title('Levenspiel plot')

plt.scatter([sol_low.root,sol_med.root,sol_hig.root],[1/rate_low,1/rate_med,1/rate_hig])
```
