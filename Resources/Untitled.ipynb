```python
# A <-> B <-> C reaction network
import numpy as np
import matplotlib.pyplot as plt
from scipy.interpolate import BPoly

kB = 8.61733e-5      # eV /K
h = 4.13566766e-15  # eV s
R0 = 8.31441 # J/mol/K

H2O = np.array([-286,-237])
N2  = np.array([0,0])
O2  = np.array([0,0])
NO  = np.array([91,88])
NO2 = np.array([33,51])
HNO3= np.array([-111,-216])

NOf = NO - 0.5*N2 - 0.5*O2
NO2f= NO2 - 0.5*N2 - O2
HNO3f = HNO3 - 0.5*H2O - 0.5*N2 - 2.5*O2

# Enthalpy surface

pH = BPoly.from_derivatives([0, 1, 2, 3], [[0, 0], [NOf[0], 0], [NO2f[0], 0], [HNO3f[0],0]])

x=np.linspace(0,3,100)
y=pH(x)

plt.figure()
plt.plot(x,y)
plt.scatter([0,1,2,3],[0,NOf[0],NO2f[0],HNO3f[0]],marker='_')

plt.annotate("air + H2O",(0,0),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("NO",(1,NOf[0]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("NO2",(2,NO2f[0]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("HNO3 (aq)",(3,HNO3f[0]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.ylabel("Enthalpy (kJ/mol HNO3 aq)")
plt.title("Standard state enthalpy at 298 K")

pH = BPoly.from_derivatives([0, 1, 2, 3], [[0, 0], [NOf[1], 0], [NO2f[1], 0], [HNO3f[1],0]])
```

```python
x=np.linspace(0,3,100)
y=pH(x)

plt.figure()
plt.plot(x,y)
plt.scatter([0,1,2,3],[0,NOf[1],NO2f[1],HNO3f[1]],marker='_')

plt.annotate("air + H2O",(0,0),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("NO",(1,NOf[1]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("NO2",(2,NO2f[1]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.annotate("HNO3 (aq)",(3,HNO3f[1]),xytext=(5, -5), textcoords='offset points', ha='left', va='bottom')
plt.ylabel("Free energy (kJ/mol HNO3 aq)")
plt.title("Standard state free energy at 298 K")
```
